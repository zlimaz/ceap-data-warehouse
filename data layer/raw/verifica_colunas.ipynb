```python
import os
import pandas as pd

# Nome do arquivo que estamos procurando
alvo = 'dirty_deputies_v2.csv'
caminho_encontrado = None

print(f" Procurando por '{alvo}' em todo o ambiente...")

# Varre todas as pastas acessíveis pelo Jupyter
for raiz, pastas, arquivos in os.walk('.'):
    if alvo in arquivos:
        caminho_encontrado = os.path.join(raiz, alvo)
        print(f"\n ACHEI! O caminho correto é:\n {caminho_encontrado}")
        break

if caminho_encontrado:
    # Tenta ler usando o caminho que o Python achou
    print("\nTentando ler com o caminho encontrado...")
    try:
        df = pd.read_csv(caminho_encontrado, nrows=5)
        print("\n SUCESSO! AS COLUNAS SÃO:")
        print("---------------------------")
        print(list(df.columns))
        print("---------------------------")
    except Exception as e:
        print(f"Erro ao ler: {e}")
else:
    print("\n NÃO ENCONTRADO.")
    print("Liste o que existe na pasta atual para debug:")
    print(os.listdir('.'))
    print("\nDICA: Se aparecer a pasta do seu projeto (ex: 'sbd2-projeto-final'), clique nela para entrar antes de criar o notebook!")
```
