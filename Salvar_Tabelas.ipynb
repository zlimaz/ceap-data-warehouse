```python
import pandas as pd
import os
from sqlalchemy import create_engine

# --- REFAZENDO A CONEXÃO (Atualizada com seus dados) ---
# Usuário: admin | Senha: admin_password | Banco: ceap_dw
db_connection_str = 'postgresql+psycopg2://admin:admin_password@localhost:5432/ceap_dw'
db_connection = create_engine(db_connection_str)

# --- INÍCIO DA EXPORTAÇÃO ---

# Criar pasta para salvar os arquivos
output_dir = 'dados_powerbi'
os.makedirs(output_dir, exist_ok=True)

print("🚀 Exportando tabelas para CSV...")

# Lista das tabelas do DW
tabelas = ['fat_rmb', 'dim_dpt', 'dim_frn', 'dim_prt', 'dim_loc', 'dim_cat', 'dim_tmp']

for tabela in tabelas:
    try:
        print(f"   -> Baixando {tabela}...")
        query = f"SELECT * FROM dw.{tabela}"
        df = pd.read_sql(query, db_connection)
        
        # Salva em CSV (separador ponto e vírgula para abrir fácil no Excel/PowerBI)
        df.to_csv(f"{output_dir}/{tabela}.csv", index=False, sep=';', encoding='utf-8-sig')
    except Exception as e:
        print(f"   ❌ Erro ao baixar {tabela}: {e}")

print(f"\n✅ Sucesso! Verifique a pasta '{output_dir}'. Copie ela para o Windows!")
```
